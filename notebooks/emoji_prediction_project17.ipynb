## Results Summary

```python
import json
import pandas as pd

metrics = json.loads(Path('results/metrics_summary.json').read_text(encoding='utf-8'))
rows = []
for model_name, values in metrics['test'].items():
    rows.append({
        'model': model_name,
        'top1_accuracy': values['top1_accuracy'],
        'top3_accuracy': values['top3_accuracy'],
        'macro_f1': values['macro_f1'],
        'weighted_f1': values['weighted_f1'],
    })
pd.DataFrame(rows)
```
